```python
def find(parent, i):
    if parent[i] != i:
        parent[i] = find(parent, parent[i])
    return parent[i]


def union(parent, rank, x, y):
    root_x = find(parent, x)
    root_y = find(parent, y)

    if root_x != root_y:
        if rank[root_x] < rank[root_y]:
            parent[root_x] = root_y
        elif rank[root_x] > rank[root_y]:
            parent[root_y] = root_x
        else:
            parent[root_y] = root_x
            rank[root_x] += 1


def kruskal(vertices, edges):
    # Sort edges according to weight
    edges.sort(key=lambda x: x[2])

    parent = list(range(vertices))
    rank = [0] * vertices

    mst = []
    total_weight = 0

    for u, v, weight in edges:
        root_u = find(parent, u)
        root_v = find(parent, v)

        # Add edge if it does not create a cycle
        if root_u != root_v:
            mst.append((u, v, weight))
            total_weight += weight
            union(parent, rank, root_u, root_v)

        # MST contains V-1 edges
        if len(mst) == vertices - 1:
            break

    return mst, total_weight


# Input
vertices = int(input("Enter number of vertices: "))
edge_count = int(input("Enter number of edges: "))

edges = []

print("Enter edges (source destination weight):")

for i in range(edge_count):
    u, v, weight = map(int, input().split())
    edges.append((u, v, weight))


# Function Call
mst, total_weight = kruskal(vertices, edges)


# Output
print("\nEdges in Minimum Spanning Tree:")
print("Edge\tWeight")

for u, v, weight in mst:
    print(f"{u} - {v}\t{weight}")

print("\nMinimum Spanning Tree Weight:", total_weight)
```
