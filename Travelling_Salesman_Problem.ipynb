```python
from functools import lru_cache


def tsp(graph):
    n = len(graph)

    @lru_cache(None)
    def dp(current, visited):
        # If all cities are visited, return to the starting city
        if visited == (1 << n) - 1:
            return graph[current][0]

        minimum_cost = float('inf')

        for city in range(n):
            # Check if city has not been visited
            if not (visited & (1 << city)):
                cost = (
                    graph[current][city]
                    + dp(city, visited | (1 << city))
                )

                minimum_cost = min(minimum_cost, cost)

        return minimum_cost

    return dp(0, 1)


# Input
n = int(input("Enter number of cities: "))

print("Enter the distance matrix:")

graph = []

for i in range(n):
    row = list(map(int, input().split()))
    graph.append(row)


# Function Call
minimum_cost = tsp(graph)

print("\nMinimum travelling cost =", minimum_cost)
```
